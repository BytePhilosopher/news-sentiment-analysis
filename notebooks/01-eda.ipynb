# Exploratory Data Analysis (EDA) for News Sentiment Analysis

In this notebook, we will perform exploratory data analysis on the news dataset to understand the data better and derive insights that will guide our modeling efforts.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set(style='whitegrid')
```

```python
# Load the dataset
data_path = '../data/raw/news_data.csv'  # Update with the correct path to your dataset
df = pd.read_csv(data_path)

# Display the first few rows of the dataset
df.head()
```

```python
# Summary statistics
df.describe(include='all')
```

```python
# Check for missing values
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Keyword analysis
from collections import Counter

# Assuming 'text' is the column containing news articles
all_words = ' '.join(df['text']).split()
word_counts = Counter(all_words)
most_common_words = word_counts.most_common(10)
most_common_words
```

```python
# Visualizations
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='sentiment')  # Assuming 'sentiment' is the target variable
plt.title('Sentiment Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Count')
plt.show()
```

## Conclusion

In this notebook, we performed exploratory data analysis on the news dataset. We examined the dataset's structure, checked for missing values, conducted keyword analysis, and visualized the sentiment distribution. These insights will help inform our modeling approach in the next notebook.
